```python
import inspect

class MiClase:
    def __init__(self, parametro1, parametro2):
        pass

firma_init = inspect.signature(MiClase.__init__)
parametros_init = firma_init.parameters

for nombre_parametro, parametro in parametros_init.items():
    if nombre_parametro != 'self':
        print("Nombre del parámetro:", nombre_parametro)
        print("Tipo del parámetro:", parametro.annotation)
        print("Valor por defecto:", parametro.default)
        print("Es requerido:", parametro.default == inspect.Parameter.empty)
        print()
```
